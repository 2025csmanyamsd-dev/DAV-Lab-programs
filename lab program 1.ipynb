```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40, 50])

print("NumPy Array:")
print(arr)

print("\n--- Array Indexing ---")
print("First element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])

print("\n--- Array Slicing ---")
print("First three elements:", arr[0:3])
print("Elements from index 2:", arr[2:])
print("Last three elements:", arr[-3:])

print("\n--- Basic Array Operations ---")
print("Add 5:", arr + 5)
print("Subtract 5:", arr - 5)
print("Multiply by 2:", arr * 2)
print("Divide by 10:", arr / 10)

print("\n--- Mathematical Operations ---")
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))

print("\n--- Array Reshaping ---")
reshaped_arr = arr.reshape(5, 1)
print("Reshaped Array:")
print(reshaped_arr)

data = {
    "Name": ["Asha", "Ravi", "Meera", "Karan"],
    "Marks": [88, 72, 91, 65],
    "Branch": ["CSE", "ECE", "CSE", "ISE"]
}

df = pd.DataFrame(data)

print("\n--- Pandas DataFrame ---")
print(df)

print("\n--- Column Indexing ---")
print("Name Column:")
print(df["Name"])

print("\nMarks Column:")
print(df["Marks"])

print("\n--- Row Indexing ---")
print("First Row:")
print(df.iloc[0])

print("\nThird Row:")
print(df.iloc[2])

print("\n--- Specific Value ---")
print("Marks of first student:", df.iloc[0, 1])

print("\n--- Row Slicing ---")
print("First two rows:")
print(df.iloc[0:2])

print("\n--- Row and Column Slicing ---")
print("First three rows with Name and Marks:")
print(df.iloc[0:3, [0, 1]])

print("\n--- Data Filtering ---")
result = df[df["Marks"] > 70]
print("Students with Marks greater than 70:")
print(result)
```
