```python
import pandas as pd
import numpy as np

# Create sample student data
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Gender": ["Male", "Female", "Male", "Female", "Male"],
    "StudyHours": [5, 6, np.nan, 4, 7],
    "Attendance": [80, 85, 75, np.nan, 90],
    "Marks": [75, 82, 68, 55, np.nan],
    "Result": ["Pass", "Pass", "Pass", "Fail", "Pass"]
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)

print("ORIGINAL DATASET")
print(df)

# Identify missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

# Mean Imputation for StudyHours
df["StudyHours"] = df["StudyHours"].fillna(
    df["StudyHours"].mean()
)

# Median Imputation for Marks
df["Marks"] = df["Marks"].fillna(
    df["Marks"].median()
)

# Forward Fill for Attendance
df["Attendance"] = df["Attendance"].ffill()

# Check missing values again
print("\nMISSING VALUES AFTER IMPUTATION")
print(df.isnull().sum())

# Encode Gender
df["Gender"] = df["Gender"].map({
    "Male": 1,
    "Female": 0
})

# Encode Result
df["Result"] = df["Result"].map({
    "Pass": 1,
    "Fail": 0
})

# Display final dataset
print("\nFINAL DATASET")
print(df)
```
